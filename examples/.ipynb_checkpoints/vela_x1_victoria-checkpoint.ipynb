```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from bliss import *
```

```python
text_path = '/Users/graciela/Desktop/git/BLiSS/examples/1928_highhardness/spec_0.dat'
energy_high, values_high, errors_high, bin_width_high = load_text_spectrum(text_path)

text_path = '/Users/graciela/Desktop/git/BLiSS/examples/1928_lowhardness/spec_0.dat'
energy_low, values_low, errors_low, bin_width_low = load_text_spectrum(text_path)
```

```python
plt.figure(figsize=(8, 4))
plt.errorbar(energy_high, values_high, yerr=errors_high, fmt=".", alpha=0.7)
plt.errorbar(energy_low, values_low, yerr=errors_low, fmt=".", alpha=0.7)
plt.xlabel("Energy (keV)")
plt.ylabel("Spectrum")
plt.title("Raw text spectrum")
plt.show()
```

```python
lines_high = find_emission_lines(
    energy_high,
    values_high,
    errors_high,
    show_plot=True,
    plot_name="xmm_bliss_fit.png",
)
```

```python
lines_low = find_emission_lines(
    energy_low,
    values_low,
    errors_low,
    show_plot=True,
    plot_name="xmm_bliss_fit.png",
)
```

```python
clean_lines_low = lines_low[(lines_low.snr_area>1) & (lines_low.snr_peak>1)]
clean_lines_high = lines_high[(lines_high.snr_area>1) & (lines_high.snr_peak>1)]


clean_lines_low=add_most_probable_ion(clean_lines_low,v_doppler_kms=1000)
clean_lines_high=add_most_probable_ion(clean_lines_high,v_doppler_kms=1000)
```

```python
fig, ax = plot_line_prob(clean_lines_high, show=False)
ax.set_title("BLiSS candidate probabilities and IDs")
plt.xscale("log")
plt.show()

fig, ax = plot_line_prob(clean_lines_low, show=False)
ax.set_title("BLiSS candidate probabilities and IDs")
plt.xscale("log")
plt.show()
```
